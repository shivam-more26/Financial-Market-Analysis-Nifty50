```python
import yfinance as yf
import pandas as pd
from google.colab import files
```

```python
tickers = {
    'Nifty50': '^NSEI',
    'Bank': '^NSEBANK',
    'IT': '^CNXIT',
    'Auto': '^CNXAUTO',
    'Pharma': '^CNXPHARMA'
}
```

```python
raw_data = yf.download(list(tickers.values()), period='1y')
df_prices = raw_data['Close']
```

```python
mapping = {v: k for k, v in tickers.items()}
df_prices.columns = [mapping.get(col,col) for col in df_prices.columns]
```

```python
df_return = df_prices.pct_change(fill_method=None) * 100
```

```python
output_filename = 'Market_Sector_Analysis.xlsx'
print(f"saving data to {output_filename}...")
with pd.ExcelWriter(output_filename) as writer:
  df_prices.to_excel(writer, sheet_name='Raw_Prices')
  df_return.to_excel(writer, sheet_name='Daily_Returns')

print("Data Pipeline complete. Downloading file now...")
files.download(output_filename)
```
